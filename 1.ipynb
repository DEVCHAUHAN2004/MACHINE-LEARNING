```python
import pandas as pd 

data = {
  'Name': ['A','B','C','D','E'],
  'Age' :[25,None,44,33,None],
  'Salary':[50000,60000,70000,None,None]

}
df = pd.DataFrame(data)
print(df)

print(df.isnull().sum())

print(df.dropna())

print(df.isnull().mean() * 100)

df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Salary'].fillna(df['Salary'].mean(),inplace=True)
print(df)
```
