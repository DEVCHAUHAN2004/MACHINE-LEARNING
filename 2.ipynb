```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

# Read CSV file
df = pd.read_csv("Sample_Data.csv")

# Create a copy of the dataframe
df_label = df.copy()

# Create LabelEncoder object
le = LabelEncoder()

# Encode categorical columns
df_label['Gender_Encoded'] = le.fit_transform(df_label['Gender'])
df_label['Passed_Encoded'] = le.fit_transform(df_label['Passed'])

# Print encoded data
print("\nLabel Encoded Data")
print(df_label[['Name', 'Gender', 'Gender_Encoded', 'Passed', 'Passed_Encoded']])


df_encoded = pd.get_dummies(df_label, columns=['City'])

print("\nOne-Hot Encoded Data (City)")
print(df_encoded)
```

```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd

# Read CSV file
df = pd.read_csv("Sample_Data.csv")

# Create a copy
df_label = df.copy()

# Label Encoding
le = LabelEncoder()

df_label['Gender_Encoded'] = le.fit_transform(df_label['Gender'])
df_label['Passed_Encoded'] = le.fit_transform(df_label['Passed'])

print("\nLabel Encoded Data")
print(df_label[['Name', 'Gender', 'Gender_Encoded', 'Passed', 'Passed_Encoded']])

# One-Hot Encoding (0 and 1)
df_encoded = pd.get_dummies(df_label, columns=['City'], dtype=int)

print("\nOne-Hot Encoded Data (City)")
print(df_encoded)
```
